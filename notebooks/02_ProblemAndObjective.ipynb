# Heart Disease Prediction Model - Problem Understanding and Objective Definition

## Problem Statement

* Heart disease is one of the leading causes of death worldwide and is strongly influenced by underlying factos such as hypertension and diabetes. Predicting heart disease at an early stage is essential for prevention as heart disease can become a serious issue very quick.

* The objective of this project is to develop a model that can analyze a patient's risk of developing heart disease by collecting, cleaning, standardizing and merging multiple publicly available datasets. By incorporating both direct and indirect risk factors, the model seeks to provide a more comprehensive and accurate assessment of cardiovascular risk. It aims to analyze hypertension and diabetes as individual factors and then give an overall probability distribution of the person's risk of heart disease.

## Objectives

* To demonstrate the application of fundamental data science techniques, including data collection, data preprocessing, exploratory data analysis, clustering, and visualization, on real-world healthcare datasets.

* To collect, clean, standardize, and integrate multiple publicly available disease-specific datasets into a unified healthcare dataset suitable for analytical tasks related to cardiovascular disease.

## Assumptions

* Publicly available datasets used in the project are assumed to be reliable and sufficiently representative for academic analysis purposes.

* Missing or inconsistent data values can be handled using standard preprocessing techniques without significantly affecting analytical outcomes.

* Disease labels provided in the original datasets are assumed to be correct and are treated as truthful indicators for analysis.

* The datasets created by integrating from multiple sources is assumed to maintain sufficient consistency after column standardization and unit normalization.

* The results of the model simply represent the risk factor of associated with each disease based on the patient's data. It is not intended for clinical diagnosis.
