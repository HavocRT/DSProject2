# Disease Prediction Model - Data Collection and Understanding

## Dataset Collection

* The datasets have been primarily collected from **Kaggle**, a popular site for reliable datasets for machine learning and data science.

* Below are the respective links for each dataset collected:

> * [Diabetes dataset](https://www.kaggle.com/datasets/iammustafatz/diabetes-prediction-dataset)
> * [Hypertension dataset](https://www.kaggle.com/datasets/miadul/hypertension-risk-prediction-dataset)
> * [Heart disease dataset](https://www.kaggle.com/datasets/oktayrdeki/heart-disease)

## Understanding the datasets

* The datasets obtained for this project from Kaggle are in a well structured and for the most part processed format. They are presented in CSV files and are easy to extract and analyze data from. The data will be converted into a uniform format in the preparation and preprocessing steps.

* ### Diabetes dataset

> * #### The diabetes dataset obtained from Kaggle is the largest dataset in the project. It contains the following features for patient records:
> > Gender, Age, Hypertension, Heart disease, Smoking history, BMI, Hemoglobin A1C level, Blood glucose level, Diabetes
>
> * Gender:
> > * Refers to the biological sex of the patient, which could have an impact on their biological susceptibility to diabetes.
> > * Values: `["Male", "Female"]`

> * Age:
> > * Age is a very important factor as diabetes is most commonly diagnosed in older adults.
> > * Values: `Ranging from 0-80`

> * Hypertension:
> > * Refers to the medical condition in which the blood pressure in the arteries is persistently elevated.
> > * Values: `[1 -> Yes, the patient has hypertension, 0 -> No, the patient does not have hypertension]`

> * Heart disease:
> > * Heart disease is another commonly diagnosed medical condition associated with people having diabetes.
> > * Values: `[1 -> Yes, the patient has a history of heart disease, 0 -> They do not have heart disease]`

> * Smoking history:
> > * Smoking history is also considered a risk factor for diabetes and can exacerbate the complications associated with diabetes.
> > * Values: `["not current", "former", "No Info", "current", "never", "ever"]`

> * BMI:
> > * Body mass index is the measure of a persons body fat based on their weight and height. Higher BMI values are linked to a higher risk of diabetes.
> > * Values: `Ranging from 10 to 72`
| BMI Range | Status |
| :--- | :--- |
|  < 18.5   | Underweight |
| 18.5 - 24.9 | Normal |
| 25 - 29.9 | Overweight |
| > 30 | Obese |

> * Hemoglobin A1C level:
> > * HbA1C level is the measure of a person's average blood sugar level over the past 2-3 months. Higher levels indicate a greater risk of developing diabetes.
> > * Values: `Ranging from 3.5 to 9`

> * Blood glucose level:
> > * Blood glucose level refers to the amount of glucose in the bloodstream at a given time. High blood glucose levels are a key indicator of diabetes.
> > * Values: `Ranging from 80 to 300`

> * Diabetes:
> > * Indicates whether the patient has diabetes or not.
> > * Values: `[1 -> Yes, 0 -> No]`

* ### Hypertension dataset

> * #### The hypertension dataset contains patient lifestyle, physiological, and medical history attributes used to analyze the risk and occurrence of high blood pressure. It contains the following features for patient records:
> > Age, Salt intake, Stress score, Blood pressure history, Sleep duration, BMI, Medication, Family history, Exercise level, Smoking status

> * Age:
> > * Refers to the age of the patient, which is an important factor since the likelihood of hypertension generally increases with age.
> > * Values: `Ranging from 18 to 84`

> * Salt intake:
> > * Represents the estimated daily dietary salt consumption, which is strongly associated with elevated blood pressure levels.
> > * Values: `Ranging from 2.5 to 16.4`

> * Stress score:
> > * Indicates the measured stress level of the individual on a standardized scale, where higher values represent greater stress that may contribute to hypertension.
> > * Values: `Ranging from 0 to 10`

> * BP history:
> > * Describes the patient's prior blood pressure classification, providing historical context for hypertension risk.
> > * Values: `["Normal", "Prehypertension", "Hypertension"]`

> * Sleep duration:
> > * Represents the average number of hours the patient sleeps per day, as insufficient or excessive sleep can influence blood pressure.
> > * Values: `Ranging from 1.5 to 11.4 hours`

> * BMI:
> > * Body Mass Index indicating body fat based on height and weight; higher BMI values are associated with increased hypertension risk.
> > * Values: `Ranging from 11.9 to 41.9`

> * Medication:
> > * Indicates the type of medication currently used for managing blood pressure or related conditions.
> > * Values: `["None", "Beta blocker", "ACE inhibitor", "Diuretic", "Other"]`

> * Family history:
> > * Indicates whether there is a family history of hypertension, which may increase genetic susceptibility.
> > * Values: `[True -> Yes, False -> No]`

> * Exercise level:
> > * Represents the overall physical activity level of the individual, which can significantly influence blood pressure control.
> > * Values: `["Low", "Moderate", "High"]`

> * Smoking status:
> > * Indicates whether the individual currently smokes, as smoking contributes to cardiovascular and blood pressure risks.
> > * Values: `["Non smoker", "Smoker"]`


* ### Heart Disease dataset

> * #### The heart disease dataset used in this project is created by combining multiple datasets to obtain a broader range of cardiovascular health indicators. It contains the following features for patient records:
> > Age, Sex/Gender, Chest pain type, Blood pressure, Cholesterol level, Fasting blood sugar, EKG results, Maximum heart rate, Exercise angina, ST depression, Slope, Number of vessels, Thalassemia result, Exercise habits, Smoking status, Family heart disease, Diabetes, BMI, High blood pressure, Heart disease diagnosis

> * Age:
> > * Refers to the age of the individual, which is an important factor since the risk of heart disease generally increases with age.
> > * Values: `Ranging from 18 to 80`

> * Sex / Gender:
> > * Indicates the biological sex of the patient, which can influence the prevalence and severity of cardiovascular diseases.
> > * Values: `[1 -> Male, 0 -> Female]` or `["Male", "Female"]`

> * Chest pain type:
> > * Represents the type of chest pain experienced by the patient, which is a key diagnostic indicator.
> > * Values: `[1 -> Typical angina, 2 -> Atypical angina, 3 -> Non-anginal pain, 4 -> Asymptomatic]`

> * Blood pressure:
> > * Resting blood pressure measured in mm Hg; elevated levels are associated with increased cardiovascular risk.
> > * Values: `Ranging from 94 to 200`

> * Cholesterol level:
> > * Serum cholesterol concentration measured in mg/dl, where higher levels are linked to increased heart disease risk.
> > * Values: `Ranging from 126 to 564`

> * Fasting blood sugar:
> > * Indicates whether fasting blood sugar exceeds 120 mg/dl, which can signal metabolic abnormalities.
> > * Values: `[1 -> Yes, 0 -> No]`

> * EKG results:
> > * Resting electrocardiographic results used to detect abnormalities in heart electrical activity.
> > * Values: `[0 -> Normal, 1 -> ST-T wave abnormality, 2 -> Left ventricular hypertrophy]`

> * Maximum heart rate:
> > * The maximum heart rate achieved during exercise testing, which helps evaluate cardiac performance.
> > * Values: `Ranging from 71 to 202`

> * Exercise angina:
> > * Indicates whether the patient experiences chest pain during exercise.
> > * Values: `[1 -> Yes, 0 -> No]`

> * ST depression:
> > * ST depression induced by exercise relative to rest, used to identify myocardial ischemia.
> > * Values: `Ranging from 0 to 6.2`

> * Slope:
> > * The slope of the peak exercise ST segment observed during stress testing.
> > * Values: `[1 -> Upsloping, 2 -> Flat, 3 -> Downsloping]`

> * Number of vessels (ca):
> > * Number of major vessels colored by fluoroscopy, indicating severity of coronary artery involvement.
> > * Values: `[0 – 3]`

> * Thalassemia result (thal):
> > * Describes thalassemia test results used in cardiovascular diagnostics.
> > * Values: `[3 -> Normal, 6 -> Fixed defect, 7 -> Reversible defect]`

> * Exercise habits:
> > * Indicates the general exercise activity level of the individual.
> > * Values: `["High", "Medium", "Low"]`

> * Smoking:
> > * Indicates whether the individual smokes.
> > * Values: `[True, False, Null]`

> * Family heart disease:
> > * Indicates whether there is a family history of heart disease.
> > * Values: `[True, False, Null]`

> * Diabetes:
> > * Indicates whether the individual has diabetes, which is a major cardiovascular risk factor.
> > * Values: `[True, False, Null]`

> * BMI:
> > * Body Mass Index representing body fat based on height and weight.
> > * Values: `Ranging from 18 to 40`

> * High blood pressure:
> > * Indicates whether the individual has clinically diagnosed high blood pressure.
> > * Values: `[True, False, Null]`

> * Heart disease diagnosis:
> > * Indicates the presence of angiographic heart disease.
> > * Values: `[0 -> < 50% diameter narrowing, 1 -> > 50% diameter narrowing]`


## Ethics and Bias Considerations

* All datasets used in this project are publicly available and anonymized, and none of these data sets contain any kind of identifiable information about individuals. However, there might be some bias in data sets, which can have affect the learning of models. Appropriate processing and normalization techniques can resolve these issues, and these data sets are used for merely academical purposes.
